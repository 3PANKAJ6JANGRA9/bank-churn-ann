```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler,LabelEncoder
import pickle
```

```python
data=pd.read_csv('Churn_Modelling.csv')
```

```python
data.head()
```

```python
data.isnull()
```

```python
data.isnull().sum()
```

```python
data.duplicated()
```

```python
data=data.drop(["RowNumber","CustomerId","Surname"], axis=1)
```

```python
data.head()
```

```python
label_encode_gender=LabelEncoder()
data['Gender']=label_encode_gender.fit_transform(data['Gender'])
```

```python
data.head()
```

```python
from sklearn.preprocessing import OneHotEncoder
one_hot_encode=OneHotEncoder(sparse_output=False)
one_hot_geo=one_hot_encode.fit_transform(data[['Geography']])
one_hot_geo
```

```python
one_hot_encode
```

```python
one_hot_encode.get_feature_names_out(['Geography'])
```

```python
encoded_geo=pd.DataFrame(one_hot_geo,columns=one_hot_encode.get_feature_names_out(['Geography']))
```

```python
encoded_geo
```

```python
data=pd.concat([data.drop('Geography',axis=1),encoded_geo],axis=1)
```

```python
data.head()
```

```python
## save encoder and scaler
with open('label_encode_gender.pkl','wb') as file:
    pickle.dump(label_encode_gender,file)

with open('one_hot_encode.pkl','wb') as file:
    pickle.dump(one_hot_encode,file)
```

```python
# Divide and split data
X=data.drop('Exited',axis=1)
y=data['Exited']

X_train,X_test,y_train,y_test=train_test_split(X,y,train_size=0.25,random_state=42)



scaler=StandardScaler()
X_train=scaler.fit_transform(X_train)
X_test=scaler.transform(X_test)
```

```python
with open('scaler.pkl','wb') as file:
    pickle.dump(scaler,file)
```

```python
#   ANN IMPLEMENTATION

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.callbacks import EarlyStopping, TensorBoard
import datetime
```

```python
#  Build the model

model=Sequential([
    Dense(64,activation='relu',input_shape=(X_train.shape[1],)), ## hidden layer 1 , declare input_shape only for HL1
    Dense(32,activation='relu'), ## HL2
    Dense(1,activation='sigmoid') ## output layer

])
```

```python
model.summary()
```

```python
# compile the model (use this way when you want to decide the learning rate)
import tensorflow as tf
opt=tf.keras.optimizers.Adam(learning_rate=0.01)
loss=tf.keras.losses.BinaryCrossentropy()
```

```python
# compile the model(use this way when you want to don't need to decide the learning rate)
optimizer=model.compile(optimizer=opt,loss="binary_crossentropy",metrics=['accuracy'])
```

```python
import datetime

# Clean format using hyphens and underscores (no slashes or colons)
log_dir = "logs/fit/" + datetime.datetime.now().strftime("%Y%m%d-%H%M%S")
```

```python
from tensorflow.keras.callbacks import EarlyStopping, TensorBoard

tensorflow_callback=TensorBoard(log_dir=log_dir,histogram_freq=1)
```

```python
#Setup Early stopping

early_stopping=EarlyStopping(monitor="val_loss",patience=10, restore_best_weights=True)
```

```python
# Train the model
history=model.fit(
    X_train,y_train ,validation_data=(X_test,y_test), epochs=100,
    callbacks=[tensorflow_callback,early_stopping]
)
```

```python
model.save('model.h5')
```
